```python
from llama_parse import LlamaParse
```

```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from pathlib import Path

file_path = Path("data/raw/Human_Resource_Policy_Manual_PRAAN.2020.pdf")
extra_info = {
    "file_name": file_path.name,
}
```

```python
parser = LlamaParse(
    api_key = os.getenv("LLAMA_PARSE_API"),
    result_type = "markdown",
)
```

```python
with open(file_path, "rb") as f:
    documents = parser.load_data(f, extra_info=extra_info)
```

```python
print(documents[15].text)
```

```python
len(documents)
```

```python
output_md_path = Path("data/processed/Human_Resource_Policy_Manual_PRAAN.2020.md")
with open(output_md_path, "w", encoding="utf-8") as f:
    for doc in documents:
        f.write(doc.text)
        f.write("\n\n")
```
